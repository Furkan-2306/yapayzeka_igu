```python
import numpy as np # kısaltılmı bir şekilde kütüphane ekler

# rasgele tanımlanan değerler - sıralaması olmayan
dizi = np.array([1, -55, 9, -23, 13, -21, 23, 37, 29, -6])


ortalama = np.mean(dizi) # tüm elemanları ortalamasını döndürür ve ekrana yazdırır
print("ortalama: ", ortalama)

std_sapma = np.std(dizi) # ne kadar yayıldığını ölçen bir değer olan standart sapmayı döndürür
print("standart sapma: ", std_sapma)


mutlak_deger= np.abs(dizi) # dizinin mutlağını alır ekrana yazdırır
print("mutlak değer = ", mutlak_deger)

```

```python
from mpl_toolkits.mplot3d import Axes3D #3d grafik oluşturmak için eklenen kütüphane

#figür nesnesi oluşturur
fig = plt.figure()
#3d eksenler için alt grafik oluşturur
#111 1 satır 1 sütun 1 alt grafiği belirler
ax = fig.add_subplot(111, projection='3d')

#-5 ile 5 arasında 100 eşit aralıklı değer üretir.
x = np.linspace(-5, 5, 100)
#-5 ile 10 arasında 50 eşit aralıklı değer üretir.
y = np.linspace(-5, 10, 50)
#x ve y dizilerini 2D bir grid (ızgara) oluşturacak şekilde genişletir.
#Böylece her bir (x, y) çifti için bir z değeri hesaplanabilir.
x, y = np.meshgrid(x,y)
#burada z, x ve y nin sinüzü hesaplanarak atanır
z = np.sin(np.sqrt(x**2 + y**2))

#verilerini kullanarak bir yüzey grafiği çizer.
#plasma renk hraitasını belirler
ax.plot_surface(x, y, z, cmap='plasma')

ax.set_title('3D Yüzey Gradiği Örneği')

plt.show()
```

```python
import matplotlib.pyplot as plt #kısaltılmış görüntü kütüphanesi
import numpy as np

#renk haritalarını göstermek için bir fonksion
def plot_colormaps():
    #mevcut tüm renk haritalarının isimlerinin lsitesi
    colormaps = plt.colormaps()
    #renk haritaların sayısı
    n = len(colormaps)
    #fig, axs = plt.subplots(n//2, 2, = bir satırdan ve iki sütundan oluşan düzen oluşturur
    #figsize=(10, 2*(n//2)) = haritaların alanalrını ve boyutlarını belirler
    fig, axs = plt.subplots(n//2, 2, figsize=(10, 2*(n//2)))

    #Her bir alt grafiğe (ax) ve renk haritası ismine (cmap_name) karşılık gelen döngü oluşturur
    for ax, cmap_name in zip(axs.ravel(), colormaps):

        cmap = plt.get_cmap(cmap_name) # renk haritasını alır

        #0 ile 1 arasında 256 değer içeren bir dizi oluşturur
        # bu diziyi 1 satır ve 256 sütunlu bir şekle getirir.
        img = np.linspace(0, 1, 256).reshape(1, -1)

        #renk haritasnı kullanrak görüntü çizer
        ax.imshow(img, aspect='auto', cmap=cmap)
        # Alt grafiğin başlığını renk haritasının ismiyle ayarlar
        ax.set_title(cmap_name, fontsize=8)
        #Eksenleri gizler, böylece sadece renk haritası görünür.
        ax.set_axis_off()
    plt.show()

plot_colormaps()
```

```python
import requests #http istekleri için kullanılır
from bs4 import BeautifulSoup #html ve xml belgelerini ayrıştırmak için kullanılır

#web sayfası url
url = 'https://github.com/dashboard'

#web sayfasını indir ve responseye kaydeder
response = requests.get(url)

#Bağlantı başarılıysa
if response.status_code == 200:

  #içeriği ayrıştırır ve html yapısına erişim sağlanır
    soup = BeautifulSoup(response.content, 'html.parser')

  #Örnek olarak sayfadaki tüm başlıkları (h1, h2, h3 vb.) çek
    for heading in soup.find_all(['h1', 'h2', 'h3']):
        print(heading.text.strip()) # başlıkların sonunu ve başındaki boşlukları temizler
else:
    print("Bağlantı başarısız: ", response.status_code)
```

```python
import requests
from bs4 import BeautifulSoup

#web sayfası url
url = 'https://github.com/Furkan-2306/yapayzeka_igu'

#Web sayfasını indir
response = requests.get(url)

#eğer bağlantı başarılıysa
if response.status_code == 200:
  #HTML içeriğini ayrıştır
    soup = BeautifulSoup(response.content, 'html.parser')

  #Sayfadaki tüm a etiketlerini bulmasını sağlar (link)
    links = soup.find_all('a')

  #Sayfadaki bütün bağlantıları bul ve yazdır
    for link in links:
        href = link.get('href') # bağlantının urlsini alır
        text = link.text.strip() # bağlantının metnini alıp baş ve sondaki boşlukları temizler
        if href:
            print("Bağlantı metni: ", text, "URL:", href)
    #Bu satır, sayfadaki toplam bağlantı sayısını yazdırır.
    print("Toplam bulunan bilgi sayısı: ", len(links), " adet")
else:
    print("Bağlantı başarısız: ", response.status_code)
```

```python
# Görüntüyü yükleme
#önce bgreden rgbeden brgye
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB)) # bgr den rgb dönüşümü yapar
plt.title("Yüklenen Görüntünün Gösterimi")
plt.show()
```

```python
#BGR formatındaki görüntüyü griye çevirir
gray_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

#bir görüntüyü (matris şeklinde) gösterir. Burada, gri tonlamalı görüntü verisi kullanılır.
plt.imshow(gray_image, cmap='gray')
plt.title("Gri Tonlanmış Görüntü")
plt.show()
```

```python
#bulanıklaştırma yapar 5 5 ne kadar bulanık olucağını gösterir
#0 standart sapmadır
blurred_image = cv2.GaussianBlur(gray_image, (5, 5), 0)

plt.imshow(blurred_image, cmap='gray')
plt.title("Bulanıklaştırılmış Görüntü")
plt.show()
```

```python
#canny kenar algılaması için kullanılır
#Bulanıklaştırma, kenar algılamada daha iyi sonuçlar elde etmek için gürültüyü azaltır.
#düşük eşik değeridir
#tüksek eşik değeri
edges = cv2.Canny(blurred_image, threshold1=100, threshold2=200)

plt.imshow(edges, cmap='gray')
plt.title("Kenar Algılama Sonucu")
plt.show()
```

```python

import cv2 #opencv
import numpy as np
import matplotlib.pyplot as plt

image1 = cv2.imread('Opencv_ornekleri/trafik-2.png')
image2 = cv2.imread('Opencv_ornekleri/trafik-3.png')

# HSV formatı, renk tespiti için daha etkilidir.
#bgr den hsv ye döbüştürür
hsv_image1 = cv2.cvtColor(image1, cv2.COLOR_BGR2HSV)
hsv_image2 = cv2.cvtColor(image2, cv2.COLOR_BGR2HSV)

#Kırmızı rengin HSV formatındaki alt ve üst sınırları tanımlanır.
#Kırmızı renk, HSV'de iki aralıkta bulunur
lower_red = np.array([0, 120, 70])
upper_red = np.array([10, 255, 255])

#belirtilen renk aralıklarını beyaz ve siyah olarak atar.
mask1_image1 = cv2.inRange(hsv_image1, lower_red, upper_red)
mask1_image2 = cv2.inRange(hsv_image2, lower_red, upper_red)

#hsv ikinci jes tanımlanır
lower_red2 = np.array([170, 120, 70])
upper_red2 = np.array([180, 255, 255])

mask2_image1 = cv2.inRange(hsv_image1, lower_red2, upper_red2)
mask2_image2 = cv2.inRange(hsv_image2, lower_red2, upper_red2)

#her iki işlem birleştirilir ve bu sayede kırmızı kısımlar açığa çıkar
red_mask_image1 = mask1_image1 + mask2_image1
red_mask_image2 = mask1_image2 + mask2_image2

#beyaz kısımları çıkartır bu sayede kırmızı kısımlar görünür
red_result_image1 = cv2.bitwise_and(image1, image1, mask=red_mask_image1)
red_result_image2 = cv2.bitwise_and(image2, image2, mask=red_mask_image2)

plt.subplot(1, 2, 1)
#rgb formatına döndürür
plt.imshow(cv2.cvtColor(red_result_image1, cv2.COLOR_BGR2RGB))
plt.title('Kırmızı Ok - Görüntü 1')

plt.subplot(1, 2, 2)
plt.imshow(cv2.cvtColor(red_result_image2, cv2.COLOR_BGR2RGB))
plt.title('Kırmızı Ok - Görüntü 2')

plt.show()
```
